````python
# 📊 分析图像-图像相似度的预测结果：正确 vs 错误样本
import pandas as pd
import os

# 读取预测结果
prediction_file = 'result/image_to_image_similarity/prediction.en.txt'
gold_file = 'all.gold.en.tmp'

# 读取预测
with open(prediction_file, 'r', encoding='utf-8') as f:
    predictions = [line.strip().split('\t') for line in f]

# 读取 gold 标签
with open(gold_file, 'r', encoding='utf-8') as f:
    gold_labels = [line.strip().split('\t')[0] for line in f]

# 读取完整结果（包含样本信息）
full_result = pd.read_csv('result/image_to_image_similarity/full_result.en.csv')

# 分析每个样本
correct_samples = []
wrong_samples = []

for idx, (pred_list, gold) in enumerate(zip(predictions, gold_labels)):
    query_image = f"{idx + 1}.jpg"  # 查询图像名称
    predicted = pred_list[0]  # 预测的第一名
    
    # 获取样本的 target phrase
    sample_info = full_result[full_result['data'] == idx].iloc[0]
    query_desc = sample_info.get('query_image', query_image)
    
    if predicted == gold:
        correct_samples.append({
            '样本ID': idx + 1,
            '查询图像': query_image,
            '正确答案': gold,
            '预测结果': predicted,
            '状态': '✅ 正确'
        })
    else:
        wrong_samples.append({
            '样本ID': idx + 1,
            '查询图像': query_image,
            '正确答案': gold,
            '预测结果': predicted,
            '状态': '❌ 错误'
        })

# 统计
total = len(predictions)
correct_count = len(correct_samples)
wrong_count = len(wrong_samples)
accuracy = correct_count / total * 100

print("=" * 80)
print(f"📊 图像-图像相似度评估结果")
print("=" * 80)
print(f"\n总样本数: {total}")
print(f"✅ 正确: {correct_count} ({accuracy:.2f}%)")
print(f"❌ 错误: {wrong_count} ({100-accuracy:.2f}%)")
print(f"准确率: {accuracy:.2f}%")

# 显示正确样本（前10个）
print("\n" + "=" * 80)
print("✅ 正确预测样本（前10个）")
print("=" * 80)
df_correct = pd.DataFrame(correct_samples[:10])
if len(df_correct) > 0:
    print(df_correct.to_string(index=False))
else:
    print("没有正确预测的样本")

# 显示错误样本（前20个）
print("\n" + "=" * 80)
print("❌ 错误预测样本（前20个）")
print("=" * 80)
df_wrong = pd.DataFrame(wrong_samples[:20])
if len(df_wrong) > 0:
    print(df_wrong.to_string(index=False))
else:
    print("没有错误预测的样本")

# 保存完整结果到文件
correct_df = pd.DataFrame(correct_samples)
wrong_df = pd.DataFrame(wrong_samples)

correct_df.to_csv('result/image_to_image_similarity/correct_samples.csv', index=False, encoding='utf-8-sig')
wrong_df.to_csv('result/image_to_image_similarity/wrong_samples.csv', index=False, encoding='utf-8-sig')

print("\n" + "=" * 80)
print("📁 完整结果已保存到:")
print(f"  ✅ result/image_to_image_similarity/correct_samples.csv ({len(correct_samples)} 个)")
print(f"  ❌ result/image_to_image_similarity/wrong_samples.csv ({len(wrong_samples)} 个)")
print("=" * 80)
````

````python
# 📁 创建错误样本的可视化文件夹
# 每个文件夹包含3张图像：查询图像、预测图像、正确答案图像
import os
import shutil
from pathlib import Path
import re

# 配置
output_base_dir = 'result/image_to_image_similarity/error_samples_visualization'
num_samples_to_visualize = 10  # 展示前10个错误样本
dataset_image_dir = 'dataset/image/test_images_resized'
query_image_dir = 'image'
label_file = 'dataset/label/en.test.data.v1.1.txt'

# 创建输出目录
os.makedirs(output_base_dir, exist_ok=True)

# 读取label文件，获取每个样本的target phrase
print("📖 读取label文件...")
target_phrases = []
with open(label_file, 'r', encoding='utf-8') as f:
    for line in f:
        parts = line.strip().split('\t')
        if len(parts) >= 2:
            # 第二列是target phrase
            phrase = parts[1].split(',')[0]  # 取逗号前的部分作为主要短语
            # 清理phrase，使其适合作为文件夹名
            phrase = re.sub(r'[<>:"/\\|?*]', '_', phrase)  # 替换Windows不允许的字符
            phrase = phrase.strip()
            target_phrases.append(phrase)

print(f"✓ 读取了 {len(target_phrases)} 个target phrases\n")

# 读取错误样本列表
if not os.path.exists('result/image_to_image_similarity/wrong_samples.csv'):
    print("⚠️ 请先运行上一个单元格生成 wrong_samples.csv")
else:
    wrong_df = pd.read_csv('result/image_to_image_similarity/wrong_samples.csv')
    
    print(f"📊 找到 {len(wrong_df)} 个错误样本")
    print(f"📁 将为前 {num_samples_to_visualize} 个错误样本创建可视化文件夹\n")
    
    # 处理前N个错误样本
    for idx in range(min(num_samples_to_visualize, len(wrong_df))):
        sample = wrong_df.iloc[idx]
        sample_id = int(sample['样本ID'])
        query_image_name = sample['查询图像']
        predicted_image = sample['预测结果']
        correct_image = sample['正确答案']
        
        # 获取对应的target phrase（样本ID从1开始，列表索引从0开始）
        phrase = target_phrases[sample_id - 1] if sample_id <= len(target_phrases) else f"sample_{sample_id}"
        
        # 创建样本文件夹（使用phrase作为文件夹名）
        folder_name = f"{sample_id:03d}_{phrase}"
        sample_folder = os.path.join(output_base_dir, folder_name)
        os.makedirs(sample_folder, exist_ok=True)
        
        print(f"[{idx+1}/{num_samples_to_visualize}] 处理样本 {sample_id} - {phrase}")
        
        # 1. 复制查询图像（原始图像）
        query_src = os.path.join(query_image_dir, query_image_name)
        query_dst = os.path.join(sample_folder, f"1_query_{query_image_name}")
        if os.path.exists(query_src):
            shutil.copy2(query_src, query_dst)
            print(f"  ✓ 查询图像: {query_image_name}")
        else:
            print(f"  ⚠️ 查询图像未找到: {query_src}")
        
        # 2. 复制预测的图像
        predicted_src = os.path.join(dataset_image_dir, predicted_image)
        predicted_dst = os.path.join(sample_folder, f"2_predicted_{predicted_image}")
        if os.path.exists(predicted_src):
            shutil.copy2(predicted_src, predicted_dst)
            print(f"  ✓ 预测图像: {predicted_image}")
        else:
            print(f"  ⚠️ 预测图像未找到: {predicted_src}")
        
        # 3. 复制正确答案图像
        correct_src = os.path.join(dataset_image_dir, correct_image)
        correct_dst = os.path.join(sample_folder, f"3_correct_{correct_image}")
        if os.path.exists(correct_src):
            shutil.copy2(correct_src, correct_dst)
            print(f"  ✓ 正确图像: {correct_image}")
        else:
            print(f"  ⚠️ 正确图像未找到: {correct_src}")
        
        # 创建说明文件
        info_file = os.path.join(sample_folder, "README.txt")
        with open(info_file, 'w', encoding='utf-8') as f:
            f.write(f"样本编号: {sample_id}\n")
            f.write(f"Target Phrase: {phrase}\n")
            f.write(f"=" * 50 + "\n\n")
            f.write(f"1️⃣ 查询图像 (Query Image):\n")
            f.write(f"   文件名: 1_query_{query_image_name}\n")
            f.write(f"   说明: 这是输入的查询图像\n\n")
            f.write(f"2️⃣ 预测图像 (Predicted Image) ❌:\n")
            f.write(f"   文件名: 2_predicted_{predicted_image}\n")
            f.write(f"   说明: 模型预测的最相似图像（错误）\n\n")
            f.write(f"3️⃣ 正确图像 (Correct Image) ✓:\n")
            f.write(f"   文件名: 3_correct_{correct_image}\n")
            f.write(f"   说明: 标注的正确答案\n")
        
        print(f"  ✓ 创建说明文件\n")
    
    print("=" * 80)
    print(f"✅ 完成！已创建 {min(num_samples_to_visualize, len(wrong_df))} 个可视化文件夹")
    print(f"📁 输出目录: {output_base_dir}")
    print("=" * 80)
    print("\n每个文件夹包含:")
    print("  1️⃣ 1_query_*.jpg - 查询图像（你的输入图像）")
    print("  2️⃣ 2_predicted_*.jpg - 预测图像（模型认为最相似的图像）❌")
    print("  3️⃣ 3_correct_*.jpg - 正确图像（真实答案）✓")
    print("  📄 README.txt - 样本说明文件")
````

## 🎯 图像-图像相似度计算

使用修改后的代码，你可以计算一张图像与其他图像的相似度

````python
# 示例1: 单张查询图像与候选图像的相似度计算
from vwsd import CLIP

# 加载 CLIP 模型
clip = CLIP('laion/CLIP-ViT-L-14-laion2B-s32B-b82K')

# 你的查询图像 (例如生成的图像)
query_image = "image/my_generated_image.jpg"

# 候选图像列表
candidate_images = [
    "dataset/image/test_images_resized/image1.jpg",
    "dataset/image/test_images_resized/image2.jpg",
    "dataset/image/test_images_resized/image3.jpg",
]

# 计算图像-图像相似度
similarity = clip.get_image_similarity(query_image, candidate_images)

# 输出结果
print("相似度分数 (越高越相似):")
for i, (cand, score) in enumerate(zip(candidate_images, similarity[0])):
    print(f"  {i+1}. {cand}: {score:.2f}")

# 找到最相似的图像
best_idx = similarity[0].index(max(similarity[0]))
print(f"\n最相似的图像: {candidate_images[best_idx]}")
````

````python
# 示例2: 批量处理 V-WSD 数据集
import os
from vwsd import data_loader, CLIP
import pandas as pd

# 加载模型和数据
clip = CLIP('laion/CLIP-ViT-L-14-laion2B-s32B-b82K')
data = data_loader('dataset')['en']

# 假设你的生成图像在 image/ 文件夹中
# 并且命名为 generated_0.jpg, generated_1.jpg, ...
generated_image_dir = "image"

results = []
for n, d in enumerate(data[:5]):  # 先测试前5个样本
    print(f"处理样本 {n+1}/5: {d['target phrase']}")
    
    # 构造查询图像路径
    query_image = os.path.join(generated_image_dir, f"generated_{n}.jpg")
    
    # 检查文件是否存在
    if not os.path.exists(query_image):
        print(f"  警告: 图像不存在 {query_image}")
        continue
    
    # 计算与所有候选图像的相似度
    similarity = clip.get_image_similarity(query_image, d['candidate images'])
    
    # 按相似度排序
    ranked = sorted(zip(similarity[0], d['candidate images']), 
                   key=lambda x: x[0], reverse=True)
    
    print(f"  最佳匹配: {os.path.basename(ranked[0][1])} (分数: {ranked[0][0]:.2f})")
    
    results.append({
        'sample_id': n,
        'target_phrase': d['target phrase'],
        'best_match': os.path.basename(ranked[0][1]),
        'score': ranked[0][0]
    })

# 显示结果
df = pd.DataFrame(results)
print("\n结果汇总:")
print(df)
````

### 核心改动说明

在 `vwsd/embedding_clip.py` 中添加了两个新方法：

1. **`get_image_embedding(images, batch_size=None)`** 
   - 只计算图像的嵌入向量
   - 返回图像的特征表示

2. **`get_image_similarity(query_images, candidate_images, batch_size=None)`**
   - 计算查询图像与候选图像之间的相似度
   - 返回相似度矩阵 (query_size × candidate_size)
   - 分数范围约为 0-100，越高越相似

### 使用场景

- **场景1**: 你用 Stable Diffusion 等模型生成了图像，想找到与生成图像最相似的候选图像
- **场景2**: 你有一张参考图像，想在候选图像集中找到最相似的图像
- **场景3**: V-WSD 任务中使用图像检索而不是文本检索

### 参数说明

**新增参数：**
- `--use-image-query`: 启用图像-图像相似度模式
- `-i, --image-dir`: 查询图像所在的目录（默认：`image`）
- `--image-pattern`: 图像文件命名模式，使用 `{n}` 表示样本编号（默认：`generated_{n}.jpg`）

**示例：**
```bash
# 如果图像命名为 query_0.png, query_1.png, ...
vwsd-clip-baseline -l en -m laion/CLIP-ViT-L-14-laion2B-s32B-b82K \
    --use-image-query \
    -i my_images \
    --image-pattern "query_{n}.png"

# 如果图像命名为 img_000.jpg, img_001.jpg, ...
vwsd-clip-baseline -l en -m laion/CLIP-ViT-L-14-laion2B-s32B-b82K \
    --use-image-query \
    -i generated_images \
    --image-pattern "img_{n:03d}.jpg"
```

**注意：**
- 图像-图像模式目前只支持英语数据集（使用 CLIP 模型）
- 图像编号从 0 开始，对应数据集中的样本顺序
- 如果某个图像文件不存在，该样本会被跳过
